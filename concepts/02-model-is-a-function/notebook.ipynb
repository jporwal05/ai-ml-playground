## Step 3 — What does the tilt knob `m` do?

**TODO:** On **one** chart, draw 4 lines with `b = 3` fixed and `m` = 0, 1, 2, 3. Give each line a label.

*Hints:*
- `x = np.linspace(0, 5, 50)` gives 50 evenly spaced x values from 0 to 5, which is smooth enough for a line.

**Question (answer in the cell as a comment):** What does `m = 0` look like? What would it mean for a taxi?

```python
# TODO
x = np.linspace(0, 5, 50)
b = 3
for m in range(0, 4):
    plt.plot(x, predict(x, m, b), label=f"m={m}")
plt.legend()
plt.show()
# m = 0 means: The rate is 0 per km and the taxi will only get the base fare or the starting fare no matter how many km's it clocks
```

## Step 5 — A negative tilt

A phone battery starts at 100% and loses 8% every hour.

**TODO:** Write the knobs (`m`, `b`) for this, plot the battery over 0–12 hours, and find when it hits 0%.

```python
# TODO
m = -8
b = 100
x = np.linspace(0, 12, 12)
plt.scatter(x, predict(x, m, b))
plt.plot(x, predict(x, m, b))
for point in x:
    plt.annotate(
        predict(point, m, b),                      # The text label you want to display
        (point, predict(point, m, b)),               # The actual (x, y) coordinate of the data point
        textcoords="offset points", # How to position the text
        xytext=(5, 5),              # Distance from the point (x offset, y offset) in points
        ha='left',                  # Horizontal alignment ('left', 'center', 'right')
        va='bottom'                 # Vertical alignment ('bottom', 'center', 'top')
    )
plt.show()

# Battery hits 0% at below x
x = 100/8
x
```
