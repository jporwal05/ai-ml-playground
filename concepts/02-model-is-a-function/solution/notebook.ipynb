```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 1 — The model as a function
def predict(x, m, b):
    return m * x + b
```

```python
# Step 2 — Taxi meter: m = 2 per km, b = 3 base fare
km = np.array([0, 1, 2, 5])
predict(km, 2, 3)   # array([ 3,  5,  7, 13])
```

```python
# Step 3 — Tilt knob m (b fixed)
x = np.linspace(0, 5, 50)
b = 3
for m in [0, 1, 2, 3]:
    plt.plot(x, predict(x, m, b), label=f"m={m}")
plt.xlabel("km")
plt.ylabel("fare ($)")
plt.legend()
plt.show()

# m = 0 means: a flat line. The fare is always the base fare, however far you go.
```

```python
# Step 4 — Shift knob b (m fixed)
m = 2
for b in [0, 3, 6]:
    plt.plot(x, predict(x, m, b), label=f"b={b}")
plt.xlabel("km")
plt.ylabel("fare ($)")
plt.legend()
plt.show()

# Each line crosses x = 0 at y = b, because m * 0 = 0 leaves only b (the base fare).
```

```python
# Step 5 — Negative tilt: battery
m, b = -8, 100
hours = np.arange(0, 13)          # 0, 1, ..., 12  (13 points)
plt.plot(hours, predict(hours, m, b), marker="o")
plt.xlabel("hours")
plt.ylabel("battery (%)")
plt.show()

# 0 = -8 * x + 100  ->  x = 100 / 8
print("battery hits 0% at", 100 / 8, "hours")
print("model at 15 h:", predict(15, m, b), "%  <- impossible: the model doesn't know reality")
```

```python
# Bonus — A line can't bend
hours = np.array([6, 8, 10, 12, 14, 16, 18])
temps = np.array([14, 18, 23, 26, 25, 21, 16])   # °C

plt.scatter(hours, temps, label="real")
plt.plot(hours, predict(hours, 0.25, 17.4), color="tab:orange", label="best possible line")
plt.xlabel("hour of day")
plt.ylabel("temperature (°C)")
plt.legend()
plt.show()

# What happened: the data rises then falls (non-linear). Any line misses badly.
```
