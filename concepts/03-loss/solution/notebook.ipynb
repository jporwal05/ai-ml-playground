```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def predict(x, m, b):
    return m * x + b
```

```python
# Step 1 — MSE: error -> squared -> mean
def mse(y, y_hat):
    err = y - y_hat
    err_squared = err ** 2
    return np.mean(err_squared)
```

```python
# Step 2 — Check against the hand-worked example
x = np.array([6, 10, 14])
y = np.array([7, 11, 15])
print("Line A:", mse(y, predict(x, 1, 0)))    # 1.0
print("Line B:", mse(y, predict(x, 0, 11)))   # 10.67
```

```python
# Step 3 — Score the Lesson 01 line, then try to beat it
sizes  = np.array([6, 8, 10, 12, 14])
prices = np.array([7.0, 9.5, 11.0, 13.8, 15.6])

for m, b in [(1.029, 1.089), (1.03, 1.09), (1.05, 0.8), (1.075, 0.63)]:
    print(f"m={m}, b={b} -> MSE {mse(prices, predict(sizes, m, b)):.4f}")
```

```python
# Step 4 — The loss curve (b fixed)
b = 1.089
ms = np.linspace(0.5, 1.5, 101)
losses = np.array([mse(prices, predict(sizes, m, b)) for m in ms])

plt.plot(ms, losses)
plt.xlabel("m")
plt.ylabel("loss (MSE)")
plt.show()

print("lowest loss at m =", ms[np.argmin(losses)])   # 1.03

# Shape: a U-shaped valley.
# m = 1.029 is just left of the bottom.
# At m = 0.7 the curve slopes down to the right -> increase m.
```

```python
# Bonus — Draw the squares
m, b = 0.6, 4
prices_hat = predict(sizes, m, b)
print("MSE:", mse(prices, prices_hat))   # 3.77

plt.scatter(sizes, prices)
plt.plot(sizes, prices_hat, color="tab:orange")
ax = plt.gca()
for x_i, y_i, y_hat_i in zip(sizes, prices, prices_hat):
    side = abs(y_i - y_hat_i)
    ax.add_patch(Rectangle((x_i, min(y_i, y_hat_i)), side, side, alpha=0.3, color="tab:red"))
ax.set_aspect("equal")
plt.show()
```
