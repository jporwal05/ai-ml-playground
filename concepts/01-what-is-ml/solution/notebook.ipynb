```python
import numpy as np
import matplotlib.pyplot as plt

sizes  = np.array([6, 8, 10, 12, 14])            # inches  (the question)
prices = np.array([7.0, 9.5, 11.0, 13.8, 15.6])  # dollars (the answer)
```

```python
# Step 1 — Look at the data
plt.scatter(sizes, prices)
plt.xlabel("Size (inches)")
plt.ylabel("Price ($)")
plt.title("Pizza size vs price")
plt.show()
```

```python
# Step 2 — The model: a line with two knobs
def predict(size, m, b):
    return m * size + b
```

```python
# Step 3 — Turn the knobs by hand
m = 1.029   # tilt
b = 1.089   # shift

plt.scatter(sizes, prices, label="real prices")
plt.plot(sizes, predict(sizes, m, b), color="tab:orange", label=f"my line (m={m}, b={b})")
plt.xlabel("Size (inches)")
plt.ylabel("Price ($)")
plt.legend()
plt.show()   # call once, after drawing everything

# My attempts (m, b) -> how it looked:
#   (1, 0)         -> whole line too low: raise b
#   (1, 1)         -> good on the left, still low on the right: raise m a little
#   (1.029, 1.089) -> close to every dot
```

```python
# Step 4 — Use the model on a pizza it never saw
predict(16, m, b)   # ≈ 17.55
```

```python
# Bonus — How wrong is the line?
predicted = predict(sizes, m, b)
for size, real, pred in zip(sizes, prices, predicted):
    print(f"{size:>2}-inch: real {real:5.2f}  predicted {pred:5.2f}  miss {real - pred:+.2f}")
```
