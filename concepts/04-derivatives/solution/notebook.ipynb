```python
import numpy as np
import matplotlib.pyplot as plt

def predict(x, m, b):
    return m * x + b

def mse(y, y_hat):
    return np.mean((y - y_hat) ** 2)
```

```python
# Step 1 — The tiny loss: one pizza, size 1, price 3
def tiny_loss(m):
    return (3 - m) ** 2

print(tiny_loss(1), tiny_loss(3))   # 4 0
```

```python
# Step 2 — Slide your foot: rise ÷ run
def slope(f, m, h=0.0001):
    rise = f(m + h) - f(m)
    return rise / h

for m in [1, 5, 3]:
    print(f"m = {m}: slope {slope(tiny_loss, m, 0.1):.1f}")   # -3.9, 4.1, 0.1
```

```python
# Step 3 — Smaller slides close in on the true slope
for h in [0.1, 0.01, 0.001, 0.0001]:
    print(f"h = {h}: slope {slope(tiny_loss, 1, h):.4f}")

# It gets closer to -4: that limit is the derivative.
```

```python
# Step 4 — Tangent lines: a rigid skateboard resting on the bowl at each spot
ms = np.linspace(0, 6, 100)
plt.plot(ms, tiny_loss(ms), label="loss")

for spot in [1, 3, 5]:
    height = tiny_loss(spot)                      # measured ONCE, at the spot
    s = slope(tiny_loss, spot)                    # measured ONCE, at the spot
    ends = np.array([spot - 0.8, spot + 0.8])     # left end, right end
    plt.plot(ends, height + s * (ends - spot), linewidth=3)   # rise = slope × run
    plt.scatter([spot], [height], zorder=3)

plt.xlabel("m")
plt.ylabel("loss")
plt.show()
```

```python
# Step 5 — The real pizza loss (all 5 pizzas, b fixed)
sizes  = np.array([6, 8, 10, 12, 14])
prices = np.array([7.0, 9.5, 11.0, 13.8, 15.6])
b = 1.089

def pizza_loss(m):
    return mse(prices, predict(sizes, m, b))

for m in [0.7, 1.03, 1.3]:
    s = slope(pizza_loss, m)
    print(f"m = {m}: slope {s:.2f} -> {'increase' if s < 0 else 'decrease'} m")

# The signs agree with the Lesson 03 curve: negative left of the bottom, positive right of it.
# At m = 1.03 the slope is small (about -0.53): nearly at the bottom, which is at m ≈ 1.0325.
print("with h = 0.01 at m = 1.03:", round(slope(pizza_loss, 1.03, 0.01), 2), "<- wrong sign: h too big")
```

```python
# Bonus — The other knob has its own slope (m fixed)
m = 1.03

def pizza_loss_b(b):
    return mse(prices, predict(sizes, m, b))

s = slope(pizza_loss_b, 1.089)
print(f"slope for b: {s:.3f} -> {'increase' if s < 0 else 'decrease'} b")
```
