## Step 4 — See the slope: draw tangent lines (skateboards 🛹)

**What it's for:** the slope −4 at m = 1 is just a number. A **tangent line** turns it into a picture.

**Picture a skateboard resting on the bowl** at one spot. The board is **straight and rigid**: it touches the bowl at that spot, has the bowl's tilt there, and does **not** bend with the bowl.

**Finding the board's two ends.** You know slope = rise ÷ run. Flip it: **rise = slope × run**. Standing on the spot m = 1 (height 4, slope −4):
- walk 0.8 right: rise = −4 × 0.8 = −3.2, so the right end is at height 4 − 3.2 = **0.8** → (1.8, 0.8)
- walk 0.8 left: rise = −4 × −0.8 = +3.2, so the left end is at height 4 + 3.2 = **7.2** → (0.2, 7.2)

**The rule:** board height at any m = **spot height + spot slope × (m − spot)**

⚠️ **The trap:** measure the height and slope **once, at the spot**. Don't ask the bowl for its height or slope at each point; if you do, your "board" bends and just traces the bowl. (See the README's skateboard section for a full picture.)
